```python
import tensorflow as tf
```

```python
rede = tf.keras.models.Sequential()
```

```python
rede.add(tf.keras.layers.Dense(
    units = 10,
    input_shape = (19,),
    activation = 'sigmoid'
))
```

```python
rede.add(tf.keras.layers.Dense(
    units = 16,
    activation = 'sigmoid'
))
```

```python
rede.add(tf.keras.layers.Dense(
    units = 6,
    activation = 'softmax'
))
```

```python
rede.compile(
    loss = 'sparse_categorical_crossentropy',
    optimizer = 'adam',
    metrics = ['accuracy']
)
```
